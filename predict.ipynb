### Процедура предсказания обученной модели LDA при поступлении нового документа. 

#### Этапы

1. Подготовка нового документа
    - токенизация;
    - лемматизация/стемминг;
     - удаление стоп-слов и нерелевантных символов.
2. Преобразование в формат, понятный модели.  
        LDA принимает на вход корпус в виде "мешка слов" (Bag-of-Words, BoW). Для этого:
    - используется обученный словарь (dictionary), который был создан во время обучения модели. Новый документ преобразуется в формат BoW с помощью этого словаря.

### Пример

```python
from gensim.models.ldamodel import LdaModel
from gensim.corpora.dictionary import Dictionary
from gensim.utils import simple_preprocess
from nltk.corpus import stopwords
import nltk
```

#### Подготовка данных


```python
nltk.download('stopwords')
stop_words = set(stopwords.words('russian'))
```

#### Функция для предобработки текста

#### Загрузка обученной модели и словаря

```python
lda_model = LdaModel.load('lda_model_6.model')  # Путь к сохраненной модели
dictionary = Dictionary.load('dictionary.dict')  # Путь к сохраненному словарю
```

#### Новый документ

```python
def preprocess_document(doc):
    tokens = simple_preprocess(doc, deacc=True, max_len=20)  # Токенизация и удаление пунктуации
    return [word for word in tokens if word not in stop_words]  # Удаление стоп-слов
```

```python
new_document = "From: lerxstere's my thing)\nSubject: WHAtp-Posting-Host: rac3.wam.umd.edu\nOrganization: University of Maryl\nLines: 15\n\n I  out there could enlighten me on this car I saw\nthe other day. It was a 2-door sports car, looked to be from the late 60s/\nearly 70s. It was called a Bricklin. The doors were really small. In addition,\nthe front bumper was separate from the rest of the body. This is \nall I know. If anyone can tellme a model name"
preprocessed_doc = preprocess_document(new_document)
```

```python
print(new_document)
print(preprocessed_doc)
```

#### Преобразование в BoW

```python
bow_vector = dictionary.doc2bow(preprocessed_doc)
```

```python
bow_vector
```

#### Предсказание темы

```python
topic_distribution = lda_model.get_document_topics(bow_vector)
```

#### Вывод результатов

```python
print("Распределение тем:")
for topic_id, prob in topic_distribution:
    print(f"Тема {topic_id}: вероятность {prob:.4f}")
```
