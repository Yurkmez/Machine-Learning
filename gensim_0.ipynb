<big><font color='lightgreen'>gensim.utils.simple_preprocess(doc, deacc=False, min_len=2, max_len=15) -</font></big>   
<u>функция для преобразования документа в список токенов в нижнем регистре, а также для игнорирования слишком коротких или слишком длинных токенов.</u>  
Она имеет следующие параметры:  
<font color='lightgreen'>doc(str) -</font>  
Относится к входному документу, к которому должна быть применена предварительная обработка.  
<font color='lightgreen'>deacc - </font>  
 (bool, необязательно)  
Этот параметр используется для удаления знаков ударения из токенов. Для этого используется функция удаления знаков ударения().  
<font color='lightgreen'>min_len - </font>  
(int, необязательно)  
С помощью этого параметра задается минимальная длина токена. Токены, длина которых меньше заданной, будут отброшены.  
<font color='lightgreen'>max_len - </font>  
(int, необязательно)  
С помощью этого параметра задается максимальнуя длина токена. Токены, длина которых превышает заданную, будут отброшены.
Результатом работы этой функции будут токены, извлеченные из входного документа.

### Импорт

```python
import numpy as np
import pandas as pd
import gensim
from gensim import corpora
from gensim.utils import simple_preprocess
from gensim import models
# _________________________________________
from smart_open import smart_open # позволяет построчно открывать и читать большие файлы из различных источников, таких как S3, HDFS, WebHDFS, HTTP или локальных и сжатых файлов.
# from pprint import pprint
import os
```

### Словарь из текста.

```python
doc = [
   "CNTK formerly known as Computational Network Toolkit",
   "is a free easy-to-use open-source commercial-grade toolkit",
   "that enable us to train deep learning algorithms to learn like the human brain."
]
```

### Словарь из файла/нескольких, сохраняем -> <font color='lightgreen'>'./data/dict_MUL.dict'.</font>

#### Получаем данные.   

##### из одного файла: doc.txt

```python
with open("./data/doc.txt", "w") as output:
    output.write(str(doc))
```

##### из нескольких файлов: first.txt, second.txt и third.txt, содержащих три строки из текстового файла (doc.txt).

```python
# определим класс с именем Read_files, в котором есть метод __iter__()
class Read_files(object):
    def __init__(self, directoryname):
        self.directoryname = directoryname
    def __iter__(self):
        for file_name in os.listdir(self.directoryname):
            for line in open(os.path.join(self.directoryname, file_name), encoding='latin'):
                yield simple_preprocess(line)
```

#### Создаем словари (dict_STF - из данных одного файла, dict_MUL - из данных 3-х файлов)

```python
# из одног файла

# Gensim считывает файл построчно, обрабатывая по одной строке за раз с помощью simple_preprocess. 
# Т.о., ему не нужно загружать весь файл в память сразу.
dict_STF = corpora.Dictionary(
 simple_preprocess(line, deacc =True) for line in open('doc.txt', encoding='utf-8')
)
```

```python
print(dict_STF)
print(dict_STF.token2id)
```

```python
# из 3-х файлов 
# (вариант 1 - с помощью созданного выше класса)
path = "data"
dict_MUL_1 = corpora.Dictionary(Read_files(path))

print(dict_MUL_1)
print(dict_MUL_1.token2id)
```

```python
# из 3-х файлов 
# (вариант 2 - поочередно добавляя данные из разных классов)
dict_MUL = corpora.Dictionary(
 simple_preprocess(line, deacc =True) for line in open('./data/first.txt', encoding='utf-8')
)
dict_MUL.add_documents(simple_preprocess(line, deacc =True) for line in open('./data/second.txt', encoding='utf-8'))
dict_MUL.add_documents(simple_preprocess(line, deacc =True) for line in open('./data/third.txt', encoding='utf-8'))
print(dict_MUL)
```

#### Сохраняем/загружаем словарь (на выходе My_dict)

```python
dict_MUL.save('./data/dict_MUL.dict')
```

```python
My_dict = corpora.Dictionary.load('./data/dict_MUL.dict')
print(My_dict)
print(My_dict.token2id)
```

### BoW (+ tfidf) - сохраняем -> <font color='lightgreen'>'./data/BoW_tfidf.mm'</font>

#### initial data -> <font color='lightgreen'>some_text</font>

```python
some_text = [
   "A survey of user opinion of computer system response time",
   "Relation of user perceived response time to error measurement",
   "The generation of random binary unordered trees",
   "The intersection graph of paths in trees",
   "Graph minors IV Widths of trees and well quasi ordering",
]
```

```python
print(type(some_text))
print(some_text[0])
```

#### some-text tokenized (split, stoplist) -> <font color='lightgreen'>some_text_tokenized</font>

```python
stoplist = set('for a of the and to in'.split(' '))
some_text_tokenized = [[word for word in document.lower().split() if word not in stoplist] for document in some_text]
print(some_text_tokenized)
print(type(some_text_tokenized))
print(some_text_tokenized[0][0])
```

#### dictionary -> <font color='lightgreen'>dict_some_text</font>

```python
dict_some_text = corpora.Dictionary(some_text_tokenized)
print(dict_some_text.token2id)
```

#### BoW -> <font color='lightgreen'>BoW_corpus_some_text</font>

```python
BoW_corpus_some_text = [dict_some_text.doc2bow(doc, allow_update=True) for doc in some_text_tokenized]
print(BoW_corpus_some_text)
```

```python
# Для наглядности можно преобразовать идентификаторы в слова
id_words = [[(dict_some_text[id], count) for id, count in line] for line in BoW_corpus_some_text]
print(id_words)
```

#### tfidf -> <font color='lightgreen'>BoW_tfidf</font>

##### Частный случай

```python
tfidf = models.TfidfModel(BoW_corpus_some_text)
words = "trees graph".lower().split()
tfidf_rezalt = tfidf[dict_some_text.doc2bow(words)]
print(dict_some_text[14])
print(dict_some_text[16])
print(tfidf_rezalt)
```

##### Общая канва

```python
tfidf = models.TfidfModel(BoW_corpus_some_text)
BoW_tfidf = tfidf[BoW_corpus_some_text]
BoW_tfidf
```

```python
print(len(BoW_tfidf))
print(len(BoW_tfidf[4]))
print(np.around(BoW_tfidf[4], decimals=2))
```

```python
for doc in tfidf[BoW_corpus_some_text]:
    print([[dict_some_text[id], np.around(freq, decimals=2)] for id, freq in doc])
```

#### Сохранение и загрузка "мешка слов"

```python
corpora.MmCorpus.serialize('./data/BoW_tfidf.mm', BoW_tfidf)
```

```python
My_BoW = corpora.MmCorpus('./data/BoW_tfidf.mm')
for line in My_BoW:
    print(line)
```

```python
print(len(My_BoW))
print(len(My_BoW[4]))
print(np.around(My_BoW[4], decimals=2))
```
