## Дано.
Станок №1 производит детали размером 64 мм в диаметре и стандартным отклонением 4 мм.
Станок №2 производит детали размером 52 мм в диаметре и стандартным отклонением в 2 мм.
Все изделия перемешаны.

## Задача.
Перебрать детали и для каждой определить вероятность того, что она была произведена на станке №1 или на станке №2. 

```python
# импорт библиотек
import numpy as np
import random
import math
from sklearn.metrics import accuracy_score
```

```python
# начальные условия
N1 = 6000 # количество изделий произведенных на станке №1
N2 = 4000 # количество изделий произведенных на станке №2
N = N1+N2 # количество изделий произведенных на обоих станках
mu_1 = 64. # диаметр изделия станка №1
sigma_1 = 3.5 # стандартное отклонение в размере диаметра изделий станка №1
mu_2 = 52 # диаметр изделия станка №2
sigma_2 = 2. # стандартное отклонение в размере диаметра изделий станка №2
```

```python
# функция определения плотности вероятности извлечь деталь, произведенную на заданном станке
# то есть мы задаем параметры станка: мат.ожидание, среднее кв. отклонение
def gaus_func_01(mu,sigma,x):
    return math.e**(-(x-mu)**2/(2*sigma**2)) / (sigma*(2*math.pi)**0.5)

# функция определения вероятностей принадлежности деталей к станку
def proba_x(X, w1, w2, mu_1, mu_2, sigma_1, sigma_2):
    for i in X:
        P1_x = gaus_func_01(mu_1,sigma_1,i)
        P2_x = gaus_func_01(mu_2,sigma_2,i)
        P_x = w1*P1_x + w2*P2_x
        P_x_1 = (w1*P1_x)/P_x
        P_x_2 = (w2*P2_x)/P_x
        proba_temp = []
        proba_temp.append(P_x_1)
        proba_temp.append(P_x_2)
        proba_X.append(proba_temp)
    return proba_X
```

```python
# функция отнесения изделия к тому или станку
def pred_x(proba_X, limit_proba):
    pred_X = []
    for x in proba_X:
        if x[0] >= limit_proba:
            pred_X.append(1)
        else:
            pred_X.append(2)
    return np.array(pred_X)
```

```python
X = np.zeros((N))
np.random.seed(seed=42)

X[:N1] = np.random.normal(loc=mu_1, scale=sigma_1, size=N1) # инициализируем данные по деталям, производства станка №1
X[N1:N] = np.random.normal(loc=mu_2, scale=sigma_2, size=N2) # инициализируем детали, производства станка №2

y = np.zeros((N)) # инициализируем вектор ответов
y[:N1] = np.array((1))
y[N1:N] = np.array((2))

w1 = float(N1)/N # определим априорную вероятность извлечь изделие, произведенное на станке №1
w2 = float(N2)/N # определим априорную вероятность извлечь изделие, произведенное на станке №2

proba_X = [] # для каждой детали определим вероятность принадлежности к тому или иному станку
proba_X = proba_x(X, w1, w2, mu_1, mu_2, sigma_1, sigma_2)
print(proba_X)
limit_proba = 0.5 # установим порог вероятности, при достижении которого, изделие будет относиться к тому или иному станку

pred_X = [] # определим принадлежность детали к станку
pred_X = pred_x(proba_X, limit_proba)

# определим качество нашего алгоритма
print('Доля верно определенных изделий:', round(accuracy_score(y, pred_X),3))
```
