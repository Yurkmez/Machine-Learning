```python
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet
from nltk import pos_tag

lemmatizer = WordNetLemmatizer()

# Функция для преобразования тегов POS в формат WordNet
def get_wordnet_pos(tag):
    tag_dict = {
        'J': wordnet.ADJ,  # прилагательное
        'N': wordnet.NOUN,  # существительное
        'V': wordnet.VERB,  # глагол
        'R': wordnet.ADV   # наречие
    }
    return tag_dict.get(tag[0].upper(), wordnet.NOUN)  # по умолчанию — существительное

# Токенизированные тексты (каждый текст — это список слов)
tokenized_texts = [
    ["Cats", "are", "running", "faster", "than", "dogs", "."],
    ["The", "children", "played", "outside", "."],
    ["Better", "solutions", "are", "often", "the", "hardest", "to", "find", "."]
]

# Лемматизация токенизированных текстов
lemmatized_texts = []
for tokens in tokenized_texts:
    # Определяем части речи для каждого слова
    pos_tags = pos_tag(tokens)
    # Лемматизация с учетом части речи
    lemmatized_words = [lemmatizer.lemmatize(word, get_wordnet_pos(tag)) for word, tag in pos_tags]
    lemmatized_texts.append(lemmatized_words)

# Вывод результатов
print("Исходные токенизированные тексты:")
for tokens in tokenized_texts:
    print(tokens)

print("\nЛемматизированные тексты:")
for tokens in lemmatized_texts:
    print(tokens)
```
