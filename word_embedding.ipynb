__Word2Vec__ Вложение слов с помощью Word2Vec включает в себя анализ изученных векторов, а также изучение векторной математики для представления слов. 

#### Импорт

```python
from gensim.models import Word2Vec
import numpy as np
import matplotlib.pyplot as pyplot
from sklearn.decomposition import PCA
```

#### Data (это уже подготовленный список списков - очищенный и, возможно, лематизированный?)

```python
sentences = [
 ['this', 'is', 'gensim', 'tutorial', 'for', 'free', 'implementing'],
 ['this', 'is', 'the', 'tutorials', 'point', 'website'],
 ['you', 'can', 'read', 'technical','tutorials', 'for','free'],
 ['we', 'are', 'implementing','word2vec'],
 ['learn', 'full', 'gensim', 'tutorial']
]
```

```python
type(sentences)
```

#### Обучение модели

```python
model = Word2Vec(sentences, min_count=1)
model
```

#### Что мы получили

```python
# К-во слов в словаре
vocab_len = len(model.wv)
vocab_len
```

```python
# Какой индекс у конкретного слова
can_idx = model.wv.key_to_index["can"]
can_idx
```

```python
# Посчитать, сколько раз конкретное слова встретилось в документе(документах)
tutorials_cnt = model.wv.get_vecattr("tutorials", "count")
tutorials_cnt
# word_count = sum(sentence.count("tutorials") for sentence in sentences)
# print("Фактическое количество:", word_count)
```

```python
# Получение нормированных векторов
vector1 = model.wv.get_vector("tutorials", norm=True)
vector2 = model.wv.get_vector("tutorial", norm=True)

# Косинусное сходство
cosine_similarity = np.dot(vector1, vector2)
print(f"Косинусное сходство: {cosine_similarity}")
```

```python
# Косинусное сходство
sim_tutorials_tutorial = model.wv.similarity("tutorials", "tutorial")
sim_tutorials_point = model.wv.similarity("tutorials", "point")

print(f"Сходство tutorials и tutorial: {sim_tutorials_tutorial}")
print(f"Сходство tutorials и point: {sim_tutorials_point}")
```

```python
# Векторы слов
vector_tutorials = model.wv["tutorials"]
vector_tutorial = model.wv["tutorial"]
vector_point = model.wv["point"]

print(f"Вектор tutorials: {vector_tutorials}")
print(f"Вектор tutorial: {vector_tutorial}")
print(f"Вектор point: {vector_point}")
```

```python
# Посмотреть матрицу эмбеддинга
embedding_matrix = model.wv.vectors
print(embedding_matrix.shape)
print(type(embedding_matrix))
print(embedding_matrix[:1])
```

```python
# Какие слова соответствуют строкам матрицы:
words = model.wv.index_to_key
print(words)  # Список всех слов в словаре
print(type(words))  # Список всех слов в словаре
```

#### Принцип работы модели

vector_tutorials и вообще __любой вектор слова__ - это вектор, размерность которого в моделях Word2Vec и других эмбеддингах определяется гиперпараметром, который задаётся при обучении модели. Размерность выбирается так, чтобы вектор мог эффективно захватывать скрытые (латентные) семантические и синтаксические свойства слов. При обучении Word2Vec модель позволяет задавать размерность с помощью параметра vector_size = .... Обычно используются значения от 50 до 300.  
Если текстовый корпус небольшой, размерность 100 или даже 50 может быть достаточной, так как информации в корпусе недостаточно, чтобы использовать более высокие размерности.  
Для больших корпусов (миллионы предложений) размерности 200–300 позволяют модели лучше захватывать сложные семантические отношения.

__Пример:__ Как размерность влияет на векторное пространство  
Представьте трёхмерное пространство, в котором каждое слово — это точка. Трёхмерное пространство (размерность 3) может быть слишком маленьким, чтобы различить все слова. Например:

Слова "king", "queen", "man", "woman" могут быть слишком близки, чтобы выразить их различия.
Если вы увеличите размерность до 100, у модели появится больше осей для описания различий, таких как:

Пол ("мужской" или "женский").
Роль ("монарх" или "обычный человек").
Количество ("единственное" или "множественное").
Контекст ("исторический", "современный").

  
В модели Word2Vec __матрица, представляющая эмбеддинги,__ имеет следующую структуру:

Размерность матрицы эмбеддингов  
__Число строк:__ соответствует количеству уникальных слов в словаре модели (вокабуляре). Каждая строка матрицы представляет вектор для одного слова.  
__Число столбцов:__ соответствует размерности эмбеддинга (например, 100, 200, 300 и т.д.). Каждая колонка описывает одно из измерений в векторном пространстве.

__Как это используется в Word2Vec?__  
<u>Инициализация:</u>
Матрица эмбеддингов обычно инициализируется случайными числами.  
<u>Обучение:</u>  
Во время обучения веса матрицы (значения) обновляются таким образом, чтобы слова с похожими контекстами имели близкие векторы.


__Как столбцы связаны с измерениями?__  
Каждый столбец представляет одно из скрытых измерений пространства эмбеддингов. Эти измерения не имеют явного смысла (например, "цвет" или "размер"), но вместе они кодируют сложные семантические и синтаксические свойства.

Например:  
Первый столбец может (грубо) кодировать, насколько слово связано с животными.  
Второй столбец может быть связан с эмоциональной окраской.  
Третий столбец может описывать что-то связанное с физическими объектами.

#### Рисуем

```python
# Получаем матрицу эмбеддингов
embeddings = model.wv.vectors
# Применяем PCA
pca = PCA(n_components=2)
result = pca.fit_transform(embeddings)
# Визуализируем результат
pyplot.figure(figsize=(10, 8))
for i, word in enumerate(words[:50]):  # Ограничиваем до первых 50 слов для наглядности
   pyplot.scatter(result[i, 0], result[i, 1])
   pyplot.text(result[i, 0], result[i, 1], word, fontsize=9)
pyplot.show()
```
