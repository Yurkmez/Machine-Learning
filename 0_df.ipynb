## loc
https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.loc.html#pandas.DataFrame.lo

```python
# loc - выборка по строкам. Работает с метками (по существу с индексами),
# которые, если надо взять несколько строк, объдиняются в список.
# Если надо ограничить вывод данных по столбцам - то синтаксис аналогичен.

# df1.loc[['a', 'b', 'd'], ['A','D']] # -------- Столбцы на выбор. 
# df1.loc[['a', 'b', 'd'], 'B':'C']   # -------- Столбцы по порядку.
df1.loc['a', 'B']                     # -------- Одно конкретное значение.
```

```python

# # Выборка по условию

# df1.loc[df1['A'] > 0, ['A', 'B']]
df1.loc[(df1['A'] > 0) & (df1['B'] < 0), ['A', 'B']]
```

```python
# Присвоение значения: первая переменная - "для кого", вторая - "кому".

# df1.loc['a', 'A'] = 0.5
df1.loc[df1['A'] == 0.5] = 0.42   # можно +0.2 и т.п. операции.
df1
 
```

## iloc
https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.iloc.html

```python
# iloc - та же выборка по строкам. Работает с порядковыми номерами строк (0, 1, ...),
# Синтаксис ":". 
# Если надо ограничить вывод данных по столбцам - то синтаксис аналогичен.

df1.iloc[0:2, :2]
```
