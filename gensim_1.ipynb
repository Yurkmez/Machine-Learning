## Transforming Documents

### Step 1: Creating the Corpus

```python
import gensim
import pprint
from collections import defaultdict
from gensim import corpora
```

```python
t_corpus = ["CNTK formerly known as Computational Network Toolkit", 
            "is a free easy-to-use open-source commercial-grade toolkit", 
            "that enable us to train deep learning algorithms to learn like the human brain.", 
            "You can find its free tutorial on tutorialspoint.com", 
            "Tutorialspoint.com also provide best technical tutorials on technologies like AI deep learning machine learning for free"]
```

```python
stoplist = set('for a of the and to in'.split(' '))
```

```python
processed_corpus = [[word for word in document.lower().split() if word not in stoplist] for document in t_corpus]
processed_corpus
```

#### Пример использования.

```python
from collections import defaultdict

text = "гость гость хозяин хозяин хозяин"
word_count = defaultdict(int)

for word in text.split():
    word_count[word] += 1

print(word_count)  # Вывод: defaultdict(<class 'int'>, {'гость': 2, 'хозяин': 3})
```

```python
from collections import defaultdict

words = ["один", "два", "три", "четыре", "пять"]
grouped_by_length = defaultdict(list)

for word in words:
    grouped_by_length[len(word)].append(word)

print(grouped_by_length)  
```

```python
from collections import defaultdict

edges = [(1, 2), (1, 3), (2, 3)]
graph = defaultdict(list)

for u, v in edges:
    graph[u].append(v)
    graph[v].append(u)

print(graph)
```

#### Дальше ...

```python
frequency = defaultdict(int)
for text in processed_corpus:
   for token in text:
        frequency[token] += 1
        
pprint.pprint(frequency)
processed_corpus = [
      [token for token in text if frequency[token] > 1] 
      for text in processed_corpus
 ]
pprint.pprint(processed_corpus)
```

```python
dictionary = corpora.Dictionary(processed_corpus)
print(dictionary)
```

```python
BoW_corpus = [dictionary.doc2bow(text) for text in processed_corpus] 
pprint.pprint(BoW_corpus)
```

Как ни странно, но мы сфомировали мешок слов из словаря, 
содержащего слова, встречающиеся в документе не менее 2-х раз.

```python
# Для наглядности можно преобразовать идентификаторы в слова
id_words = [[(dictionary[id], count) for id, count in line] for line in BoW_corpus]
pprint.pprint(id_words)
```

### Step 2: Creating a Transformation

```python
from gensim import models
import numpy as np
```

```python
tfidf = models.TfidfModel(BoW_corpus, smartirs='ntc')
```

### Step 3: Transforming Vectors

```python
corpus_tfidf = tfidf[BoW_corpus]
for doc in corpus_tfidf:
   print(np.round(doc, decimals = 2))
```

```python
for doc in corpus_tfidf:
 print( [[dictionary[id], np.around(freq, decimals=2)] for id, freq in doc])
```
